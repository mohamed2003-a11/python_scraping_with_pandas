```python

#Classification	BMI range - kg/m2:

# Severe Thinness	< 16
# Moderate Thinness	16 - 17
# Mild Thinness	17 - 18.5
# Normal	18.5 - 25
# Overweight	25 - 30
# Obese Class I	30 - 35
# Obese Class II	35 - 40
# Obese Class III	> 40
```

```python
# 1. Ask for inputs first
name =  input("Please enter your name: ")
Weight = float(input("Please enter your weight in pounds: "))
Hight = float(input("Please enter your height in inches: "))

# 2. Calculate BMI
BMI = (Weight * 703) / (Hight * Hight)

# 3. Format and evaluate results
if BMI < 16:
    print(f"Well {name}! Your BMI is {BMI:.2f}, which is Severe Thinness")
elif BMI <= 17:
    print(f"Well {name}! Your BMI is {BMI:.2f}, which is Moderate Thinness")
elif BMI <= 18.5:
    print(f"Well {name}! Your BMI is {BMI:.2f}, which is Mild Thinness")
elif BMI <= 25:
    print(f"W ell {name}! Your BMI is {BMI:.2f}, which is Normal")
elif BMI <= 30:
    print(f"Well {name}! Your BMI is {BMI:.2f}, which is Overweight")
else:
    print(f"Well {name}! Your BMI is {BMI:.2f}, which is Obese")
 
```
